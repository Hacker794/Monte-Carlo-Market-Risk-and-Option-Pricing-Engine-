```python
import numpy as np 
import matplotlib.pyplot as plt

S0 = 100
mu = 0.07
sigma = 0.2
T = 1
n_steps = 252
dt = T / n_steps
```

```python
z = np.random.normal(0, 1, size=n_steps)

print(z.shape)
```

```python
growth_factors = np.exp(
    (mu - 0.5 * sigma**2) * dt + sigma * np.sqrt(dt) * z
)

print(growth_factors)
```

```python
prices = S0 * np.cumprod(growth_factors)

prices = np.insert(prices, 0, S0)

print(prices.shape)
```

```python
plt.plot(prices)
plt.xlabel("Trading Day")
plt.ylabel("Stock Price")
plt.title("Simulated GBM Stock Price Path")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

S0 = 100
mu = 0.08
sigma = 0.20
T = 1
n_steps = 252
dt = T / n_steps

for i in range(10000):
    z = np.random.normal(0, 1, size=n_steps)

    growth_factors = np.exp(
        (mu - 0.5 * sigma**2) * dt
        + sigma * np.sqrt(dt) * z
    )

    prices = S0 * np.cumprod(growth_factors)
    prices = np.insert(prices, 0, S0)

    plt.plot(prices)

plt.xlabel("Trading Day")
plt.ylabel("Stock Price")
plt.title("10,000 Simulated GBM Paths")
plt.show()
```
