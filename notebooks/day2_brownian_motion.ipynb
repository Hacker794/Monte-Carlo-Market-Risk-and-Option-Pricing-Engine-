```python
import numpy as np

S = 100
mu = 0.08
sigma = 0.20
dt = 1 / 252

z = np.random.normal(0, 1)

next_price = S * np.exp(
    (mu - 0.5 * sigma**2) * dt
    + sigma * np.sqrt(dt) * z
)

print("Random shock:", z)
print("Next price:", next_price)
```

```python
for i in range(100):
    z = np.random.normal(0, 1)

    next_price = S * np.exp(
        (mu - 0.5 * sigma**2) * dt
        + sigma * np.sqrt(dt) * z
    )

    print(
        "Shock:", round(z, 3),
        "Next price:", round(next_price, 2)
    )
```
