```python
# Starting with fake stock prices

import numpy as np
import pandas as pd

prices = np.array([100, 102, 101, 105, 103])

print(prices)
```

```python
# Calculating the simple returns

simple_returns = prices[1:] / prices[:-1] - 1

print(simple_returns)
```

```python
# Calculating log returns

log_returns = np.log(prices[1:] / prices[:-1])

print(log_returns)
```

```python
# Putting results in a table

df = pd.DataFrame({
    "Price": prices
})

df["Simple Return"] = df["Price"].pct_change()

df["Log Return"] = np.log(df["Price"] / df["Price"].shift(1))

print(df)
```
